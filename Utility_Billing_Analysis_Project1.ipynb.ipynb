```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sqlite3
```

```python
Utility_Data = {
    "Customer_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    "Bill_Amount": [1150, 1500, 2000, 1400, 1700, 1800, 1600, 1650, 1900, 2500],
    "Payment_Status": ["Paid", "Unpaid", "Pending", "Null", "Paid", "Unpaid", "Pending", "Null", "Paid", "Unpaid"]
}
```

```python
df = pd.DataFrame(Utility_Data)
```

```python
df
```

```python
conn=sqlite3.connect("utility.db")
```

```python
df.to_sql("billing_table",conn,if_exists="replace",index=False)
```
